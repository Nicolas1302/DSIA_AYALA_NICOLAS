### Ejercicio 2 – Dataset simple con Pandas

```
# Crear el diccionario con la información
data = {
    "Estudiante": ["Ana", "Juan", "Carla", "Pedro", "Lucia"],
    "Nota": [8, 6, 9, 5, 7]
}

# Crear el DataFrame de Pandas
df = pd.DataFrame(data)

# Mostrar el DataFrame
print("DataFrame de estudiantes:")
print(df)
print("-" * 20)

# Calcular y mostrar las métricas
print("Promedio de las notas:", df["Nota"].mean())
print("Nota máxima:", df["Nota"].max())
print("Nota mínima:", df["Nota"].min())
```

1\) Leer un archivo CSV con datos de encuestas
pandas → Porque ofrece la función read\_csv\(\), diseñada específicamente para importar datos tabulares desde archivos e integrarlos en DataFrames, facilitando su limpieza y análisis posterior\.

2\) Crear un gráfico de distribución de datos

3\) Realizar un modelo de Machine Learning

numpy → Porque su estructura principal, el "array" \(arreglo n\-dimensional\), está programado a bajo nivel \(en C\) y optimizado para realizar operaciones matemáticas complejas sobre vectores y matrices de forma extremadamente rápida\.
