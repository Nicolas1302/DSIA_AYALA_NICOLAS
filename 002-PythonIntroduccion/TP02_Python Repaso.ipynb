### Ejercicio 1 – Variables y operaciones

```
# Variables personales
nombre = "Nicolás"
edad = 34
ciudad = "Buenos Aires"

# Mensaje formateado
print(f"Hola, mi nombre es {nombre}, tengo {edad} años y vivo en {ciudad}.")

# Operaciones matemáticas
a = 20
b = 5

print("Suma:", a + b)
print("Resta:", a - b)
print("Multiplicación:", a * b)
print("División:", a / b)
```

### Ejercicio 2 – Condicionales

```
# Pede edad al usuario
edad_ingresada = int(input("Ingresá tu edad: "))

# Condicional para verificar si es mayor o menor
if edad_ingresada >= 18:
    print("Eres mayor de edad")
else:
    print("Eres menor de edad")
```

### Ejercicio 3 – Listas y bucles

```
# Lista de 5 números
numeros = [10, 20, 30, 40, 50]

# Variables para guardar los cálculos
suma = 0

# Bucle for
for num in numeros:
    print("Número:", num)
    suma = suma + num

# Calcular promedio
promedio = suma / len(numeros)

# Mostrar resultados finales
print("La suma total es:", suma)
print("El promedio es:", promedio)
```
